```python
import numpy as np
```

```python
arr = np.array([1,2,3,4])

print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr, 3))
```

```python
#log

arr = np.array([1,2,3,4])

print(np.log(arr))
print(np.log2(arr))
```

```python
#rounding

print(np.round(2.45))
print(np.ceil(2.45))
print(np.floor(2.45))
print(np.trunc(2.45))



arr = np.array([1,2,3,4,2,1,])
print(np.unique(arr))

print(np.sort(arr))
# -5 -3 -4 ai sob value possitive hoye jabe
print(np.abs(arr))
```
