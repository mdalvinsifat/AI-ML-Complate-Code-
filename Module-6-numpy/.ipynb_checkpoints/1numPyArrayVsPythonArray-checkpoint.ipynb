```python
import time
import numpy as np
```

```python

size = 10_000_000

pylist = list(range(size))
startTime= time.time()
sq_list = [x**2 for x in pylist]
endTime = time.time()
print(f"My Python List is {endTime - startTime} seconds")



#num py list time 

NumPylist = np.array(size)
startTime = time.time()
sq_list = NumPylist ** 2
endtime = time.time()
print(f"num py time list {endtime-startTime}")
```
