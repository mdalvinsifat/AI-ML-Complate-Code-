```python
import numpy as np
```

```python
    
arr = np.array([1,2,3,4])
    
print(np.sum(arr))

# sob value re gun korle ja ashbe 
print(np.prod(arr))
print(np.min(arr))
print(np.max(arr))

# minimum value index kore 
print(np.argmin(arr))

# maximum value index kore 
print(np.argmax(arr))

print(np.mean(arr))
print(np.std(arr))

print(np.median(arr))


print(np.var(arr))
```
